# 1. What is NumPy, and why is it widely used in Python?

  - NumPy is a fundamental library for scientific computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.


  - It is widely used because it provides fast and efficient array operations (vectorized operations) , which are significantly faster than standard Python lists, making it essential for data science, machine learning, and numerical analysis.

# 2. How does broadcasting work in NumPy?

Broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations. It enables NumPy to perform operations on arrays of different shapes by conceptually "stretching" or duplicating the smaller array so that it has a compatible shape with the larger array, without actually making copies of the data.

# 3. What is a Pandas DataFrame?

A Pandas DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure. It is the primary data structure in the Pandas library, typically used to represent data with labeled rows (index) and labeled columns.

# 4. Explain the use of the groupby() method in Pandas.

The groupby() method is used to split the DataFrame into groups based on some criterion (e.g., a specific column's unique values). After grouping, an aggregation function (like sum(), mean(), count()) is typically applied to calculate a statistic for each group, following the "split-apply-combine" strategy.

# 5. Why is Seaborn preferred for statistical visualizations?

Seaborn is preferred because it is a high-level, declarative library built on top of Matplotlib. It offers a simplified interface for creating visually appealing and informative statistical graphics. It includes specialized plots for complex visualizations like distribution, relationship, and category plots, and automatically handles statistical estimation.

# 6. What are the differences between NumPy arrays and Python lists?

The main differences between NumPy arrays (specifically the ndarray) and standard Python lists revolve around performance, data type homogeneity, and functionality for numerical computing.

NumPy Array (ndarray)-

  - Data Type- Stores elements of a single data type (homogeneous). This simplicity allows for optimized memory storage.
  - Speed & Efficiency -Much faster for numerical operations, especially with large datasets, due to vectorized operations and implementation in C. They are also more memory-efficient.
  - Functionality - Supports advanced mathematical and array operations (e.g., broadcasting, linear algebra) natively.
  - Memory Storage - Stores elements contiguously in memory.
  - Slicing Speed - Array slicing is generally faster due to contiguous memory storage.


Python List-

  - Data Type - Stores elements of different data types (heterogeneous).
  - Speed & Efficiency - Slower and less efficient for large-scale numerical computations because operations require Python loops.
  - Functionality - A basic container; has limited direct mathematical operations on the entire structure.
  - Memory Storage - Stores pointers to elements that can be scattered in memory.
  - Slicing Speed - List slicing is generally slower.


# 7. What is a heatmap, and when should it be used?

  - A heatmap is a graphical representation of data where the individual values in a matrix are represented as colors.

  - It should be used to visualize magnitude or intensity in a 2D format. A common use case is to visualize correlation matrices (where color intensity represents the strength of the correlation) or to display frequency/density data across two categories.

# 8. What does the term “vectorized operation” mean in NumPy?

A vectorized operation in NumPy refers to applying an operation to an entire array at once, rather than iterating through its elements using a Python loop. This is typically implemented in optimized C code, making these operations much faster and more efficient than their pure Python equivalents.

# 9.  How does Matplotlib differ from Plotly?

The key differences between Matplotlib and Plotly lie in their interactivity, complexity (level of abstraction), and deployment focus.

Matplotlib-

  - Interactivity - Primarily generates static images.
  - Level - Low-level library; requires more code for complex plots.
  - Deployment - Often used for publication-quality static graphics.

Plotly-

  - Interactivity- Generates interactive, web-based visualizations.
  - Level - High-level library; easier to create complex plots with less code.
  - Deployment - Ideal for web applications and dashboards due to its interactivity.

# 10. What is the significance of hierarchical indexing in Pandas?

Hierarchical indexing (also known as MultiIndex) allows a DataFrame or Series to have multiple index levels on an axis. Its significance lies in enabling the representation of higher-dimensional data in a familiar 2D structure, making it possible to work with complex data relationships and perform sophisticated data selection and reshaping.

# 11. What is the role of Seaborn’s pairplot() function?

The pairplot() function creates a matrix of plots (often a grid) where all the pairwise relationships between numerical columns in a DataFrame are visualized. It typically plots histograms/KDEs on the diagonal (for single variable distribution) and scatter plots on the off-diagonal (for bivariate relationships).

# 12. What is the purpose of the describe() function in Pandas?

The describe() function is used to generate descriptive statistics of the DataFrame's numerical columns. The output includes statistics like count, mean, standard deviation, minimum, maximum, and the quartiles (25th, 50th, 75th percentiles).

# 13. Why is handling missing data important in Pandas?

Handling missing data (e.g., represented as NaN) is crucial because most data analysis and machine learning algorithms cannot directly handle missing values. Unhandled missing data can lead to skewed results, biased conclusions, and errors in modeling. Pandas provides tools for identifying, removing, or imputing (filling) missing values.

# 17.  Explain the difference between apply() and map() in Pandas.

  - The apply() method is generally used to apply a function along an axis (row or column) of a DataFrame. It is more versatile and can handle functions that return a Series or scalar.

  - The map() method is strictly a Series method used to substitute each value in a Series with another value, typically based on a dictionary or a function. It is ideal for element-wise transformation of a single column.

# 21. Why is NumPy’s array slicing faster than Python’s list slicing?

NumPy arrays store elements contiguously in memory, which allows for very fast and direct access to data subsets (slices). Python lists store pointers to elements scattered in memory, requiring more overhead and indirections for slicing.

# 1. How do you create a 2D NumPy array and calculate the sum of each row?

```python
import numpy as np

# Create a 2D NumPy array
array_2d = np.array([[10, 5, 2],
                     [4, 8, 1],
                     [6, 3, 7]])

# Calculate the sum of each row (axis=1)
row_sums = array_2d.sum(axis=1)

print("2D Array:\n", array_2d)
print("\nSum of each row:", row_sums)
# Output:
# 2D Array:
#  [[10  5  2]
#   [ 4  8  1]
#   [ 6  3  7]]
# Sum of each row: [17 13 16]
```

# 2. Write a Pandas script to find the mean of a specific column in a DataFrame.

```python
import pandas as pd

# Create a sample DataFrame
data = {'Name': ['A', 'B', 'C', 'D'],
        'Age': [25, 30, 22, 35],
        'Score': [85.5, 90.2, 78.9, 92.1]}
df = pd.DataFrame(data)

# Specify the column
column_name = 'Score'

# Calculate the mean of the specified column
mean_score = df[column_name].mean()

print("DataFrame:\n", df)
print(f"\nMean of the '{column_name}' column: {mean_score:.2f}")
# Output:
# Mean of the 'Score' column: 86.68
```

# 4. How do you calculate the correlation matrix using Seaborn and visualize it with a heatmap?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create a sample DataFrame
data = {'A': [1, 2, 3, 4, 5],
        'B': [5, 4, 3, 2, 1],
        'C': [2, 4, 6, 8, 10]}
df = pd.DataFrame(data)

# 1. Calculate the correlation matrix
correlation_matrix = df.corr()

# 2. Visualize it with a heatmap using Seaborn
plt.figure(figsize=(6, 5))
sns.heatmap(correlation_matrix,
            annot=True, # Show correlation values on the map
            cmap='coolwarm', # Color map
            fmt=".2f", # Format the annotations
            linewidths=.5, # Lines between cells
            cbar_kws={'label': 'Correlation Coefficient'})
plt.title('Seaborn Heatmap of Correlation Matrix')
plt.show()
```

# 7. Write a program to perform element-wise multiplication of two NumPy arrays.

```python
import numpy as np

# Create two NumPy arrays
array1 = np.array([[1, 2], [3, 4]])
array2 = np.array([[5, 6], [7, 8]])

# Perform element-wise multiplication (Hadamard product)
element_wise_product = array1 * array2

print("Array 1:\n", array1)
print("\nArray 2:\n", array2)
print("\nElement-wise Product:\n", element_wise_product)
# Output:
# Element-wise Product:
#  [[ 5 12]
#   [21 32]]
```

# 8. Create a line plot with multiple lines using Matplotlib.

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x) * np.cos(x)

# Create a line plot with multiple lines
plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='sin(x)', color='blue')
plt.plot(x, y2, label='cos(x)', color='red', linestyle='--')
plt.plot(x, y3, label='sin(x)cos(x)', color='green', linestyle=':')

plt.title('Matplotlib Line Plot with Multiple Lines')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()
plt.grid(True)
plt.show()
```

# 9. Generate a Pandas DataFrame and filter rows where a column value is greater than a threshold.

```python
import pandas as pd

# Create a sample DataFrame
data = {'ID': [1, 2, 3, 4, 5, 6],
        'Sales': [50, 150, 90, 210, 80, 180]}
df = pd.DataFrame(data)

# Define the threshold
threshold = 100
column_to_filter = 'Sales'

# Filter rows where 'Sales' is greater than the threshold
filtered_df = df[df[column_to_filter] > threshold]

print("Original DataFrame:\n", df)
print(f"\nFiltered DataFrame (where {column_to_filter} > {threshold}):\n", filtered_df)
# Output:
# Filtered DataFrame (where Sales > 100):
#    ID  Sales
# 1   2    150
# 3   4    210
# 5   6    180
```

# 10. Create a histogram using Seaborn to visualize a distribution.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Sample data (a random distribution)
data = np.random.randn(1000) # 1000 points from a standard normal distribution
df = pd.DataFrame({'Value': data})

# Create a histogram using Seaborn
plt.figure(figsize=(8, 5))
sns.histplot(df['Value'],
             kde=True, # Add a Kernel Density Estimate line
             bins=30, # Number of bins
             color='skyblue')
plt.title('Seaborn Histogram with KDE for Value Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

# 11. Perform matrix multiplication using NumPy.

```python
import numpy as np

# Create two matrices (for multiplication, matrix A columns must equal matrix B rows)
matrix_a = np.array([[1, 2],
                     [3, 4]])

matrix_b = np.array([[5, 6],
                     [7, 8]])

# Perform matrix multiplication using the @ operator or np.dot()
matrix_product = matrix_a @ matrix_b
# Equivalent to: matrix_product = np.dot(matrix_a, matrix_b)

print("Matrix A:\n", matrix_a)
print("\nMatrix B:\n", matrix_b)
print("\nMatrix Multiplication (A @ B):\n", matrix_product)
# Output:
# Matrix Multiplication (A @ B):
#  [[19 22]  (1*5 + 2*7 = 19; 1*6 + 2*8 = 22)
#   [43 50]]  (3*5 + 4*7 = 43; 3*6 + 4*8 = 50)
```
